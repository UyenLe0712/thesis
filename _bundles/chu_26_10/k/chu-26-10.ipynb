```python
import subprocess, sys, os, glob, shutil
def sh(c):
    r = subprocess.run(c, shell=True, capture_output=True, text=True); print(r.stdout[-1500:], r.stderr[-800:], flush=True); return r
sh("find /kaggle/input -maxdepth 4 | head -30")
sh("java -version")
```

```python
# dựng cây ROOT/harness + ROOT/runs từ dataset (zip tự giải hoặc chưa)
W = "/kaggle/working/w"; shutil.rmtree(W, ignore_errors=True); os.makedirs(W)
src = glob.glob("/kaggle/input/*/chu-26-10") + glob.glob("/kaggle/input/chu-26-10") + glob.glob("/kaggle/input/datasets/*/chu-26-10")
print(src); S = src[0]
for z in glob.glob(S + "/*.zip"): sh(f"cd {W} && unzip -o -q {z} -d {W}/{os.path.basename(z)[:-4]}")
for d in ("harness", "runs"):
    if os.path.isdir(f"{S}/{d}"): shutil.copytree(f"{S}/{d}", f"{W}/{d}", dirs_exist_ok=True)
sh(f"find {W} -type f | head -30")
```

```python
# harness/text_metrics_them.py + chu_26_10.py nằm ở W/harness; ROOT = W
import subprocess
p = subprocess.Popen(f"cd {W}/harness && {sys.executable} -u chu_26_10.py > {W}/chu.log 2>&1", shell=True)
import time
while p.poll() is None:
    time.sleep(120); print(time.strftime("%H:%M"), open(f"{W}/chu.log").read()[-400:].replace("\n", " | "), flush=True)
print(open(f"{W}/chu.log").read()[-3000:])
```

```python
import zipfile
with zipfile.ZipFile("/kaggle/working/chu_26_10.zip", "w") as z:
    for f in ("chu.log",): z.write(f"{W}/{f}", f)
    for f in glob.glob(f"{W}/runs/text_metrics_26_10*.json"): z.write(f, os.path.basename(f))
shutil.copy(f"{W}/chu.log", "/kaggle/working/chu.log")
for f in glob.glob(f"{W}/runs/text_metrics_26_10*.json"): shutil.copy(f, "/kaggle/working/")
shutil.rmtree(W)
print(os.listdir("/kaggle/working"))
```
